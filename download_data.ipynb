```python
from datasets import load_dataset
import json

target_size = int(1.5 * 1024**3)
total_size = 0

dataset = load_dataset(
    "HuggingFaceFW/fineweb-edu",
    "sample-10BT",
    split="train",
    streaming=True
)

with open("fineweb_edu.jsonl", "w", encoding="utf-8") as f:
    for row in dataset:
        text = row["text"].strip()

        if len(text) < 200:
            continue

        line = json.dumps({"text": text}, ensure_ascii=False) + "\n"
        size = len(line.encode("utf-8"))

        if total_size + size > target_size:
            break

        f.write(line)
        total_size += size

print(f"Saved {total_size / 1024**3:.2f} GiB")
```
